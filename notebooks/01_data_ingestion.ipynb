```python
import pandas as pd
```

```python
import requests

url = "https://api.mfapi.in/mf/125497"

response = requests.get(url)

print(response.status_code)
```

```python
data = response.json()

print(data.keys())
```

```python
nav_df = pd.DataFrame(data['data'])

nav_df.head()
```

```python
nav_df.to_csv(
    "../data/raw/HDFC_Top100_NAV.csv",
    index=False
)

print("File Saved Successfully")
```
